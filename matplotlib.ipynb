```python
#importing matplotlib library
import matplotlib.pyplot as plt
```

```python
#import numpy to get data for our plots
import numpy as np
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)
z=np.cos(x)
```

```python
print(x)
print(y)
print(z)
```

```python
#plotting the data
#sin wave
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,z)
plt.show()
```

```python
#adding title,x-axis and y-xis
plt.plot(x,y)
plt.xlabel('angle')
plt.ylabel('sinevalue')
plt.title('sine wave')
plt.show()
```

```python
#parabola
x=np.linspace(-10,10,20)
y=x**2
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,'r+')
plt.plot(x,y,'g.')
plt.plot(x,y,'rx')
plt.show()
```

```python
x=np.linspace(-5,5,50)
plt.plot(x,np.sin(x),'g-')
plt.plot(x,np.cos(x),'r--')
plt.show()
```

```python
#bar plot
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
languages=['eng','french','spanish','latin']
people=[100,200,300,444]
ax.bar(languages,people)
plt.xlabel('LANGUAGES')
plt.ylabel('NO. OF PEOPLE')
plt.show()
```

```python
#pie chart
fig1=plt.figure()
ax=fig1.add_axes([0,0,1,1])
languages=['eng','french','spanish','latin']
people=[100,200,300,444]
ax.pie(people,labels=languages,autopct='%1.1f%%')
plt.show()
```

```python
#scatter plot
x=np.linspace(0,10,30)
y=np.sin(x)
z=np.cos(x)
fig2=plt.figure()
ax=fig2.add_axes([0,0,1,1])
ax.scatter(x,y,color='g')
ax.scatter(x,z,color='b')
plt.show()
```

```python
#3d scatter plot
fig3=plt.figure()
ax=plt.axes(projection='3d')
z=20*np.random.random(100)
x=np.sin(z)
y=np.cos(z)
ax.scatter(x,y,z,c=z,cmap='Blues')
plt.show()
```
